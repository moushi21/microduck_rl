```python
import os
import shutil

# 編集対象のファイルパス
target_files = [
    'src/mjlab_microduck/robot/microduck/scene.xml',
    'src/mjlab_microduck/robot/microduck/scene_ball.xml',
    'src/mjlab_microduck/robot/microduck/scene_walk.xml'
]

# 挿入する設定内容
replacement_visual = """    <visual>
        <headlight diffuse="0.6 0.6 0.6" ambient="0.3 0.3 0.3" specular="0 0 0" />
        <quality shadowsize="4096"/>
        <map shadowbias="0.008" shadowclip="0.5"/>
    </visual>"""

for file_path in target_files:
    if os.path.exists(file_path):
        # .bak 拡張子を末尾に付加してバックアップを作成
        backup_path = file_path + ".bak"
        shutil.copy2(file_path, backup_path)
        print(f"バックアップを作成しました: {backup_path}")

        with open(file_path, 'r') as f:
            content = f.read()

        # 古い <visual> ~ </visual> ブロックを新しい高品質シャドウ設定に置換
        import re
        new_content = re.sub(r'\s*<visual>.*?</visual>', f'\n{replacement_visual}', content, flags=re.DOTALL)

        with open(file_path, 'w') as f:
            f.write(new_content)
        print(f"{file_path} のシャドウ品質を向上させ、チラつき防止設定を書き込みました。\n")
    else:
        print(f"{file_path} が見つかりませんでした。\n")
```
